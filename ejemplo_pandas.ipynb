# **Ejemplo práctico con un dataset real**

Usaremos un dataset de jugadores de la NBA para todos los ejemplos

```python
url = "https://raw.githubusercontent.com/fralfaro/MAT281/main/docs/lectures/data_manipulation/data/player_info.csv"
df = pd.read_csv(url)
```

**Exploración básica del DataFrame**

```python
# Ver las primeras 5 filas
df.head()
```

```python
# Ver las últimas 5 filas
df.tail()
```

```python
# Ver las primeras n filas
df.head(10)
```

**Dimensiones y estructura**

```python
# Dimensiones (filas, columnas)
print(df.shape)

# Nombres de las columnas
print(df.columns)

# Tipos de datos por columna
print(df.dtypes)

# Información completa del DataFrame
df.info()
```

**Estadísticas descriptivas**

```python
# Resumen estadístico de columnas numéricas
df.describe()

# Incluir todas las columnas (incluidas categóricas)
df.describe(include='all')
```

**Valores únicos y frecuencias**

```python
# Frecuencia de valores en una columna categórica
df['position'].value_counts()

# Número de valores únicos
df['position'].nunique()

# Valores únicos
df['position'].unique()
```

**Selección y filtrado de datos**

Seleccionar columnas

```python
# Una columna (devuelve una Serie)
df['name']

# Múltiples columnas (devuelve un DataFrame)
df[['name', 'position', 'height']]
```

Filtrar filas por condición

```python
def convert_height_to_inches(height_str):
    if pd.isna(height_str):
        return np.nan
    try:
        feet, inches = map(int, height_str.split('-'))
        return (feet * 12) + inches
    except ValueError:
        return np.nan

df['height_inches'] = df['height'].apply(convert_height_to_inches)

# Jugadores más altos de 2 metros (height está en pulgadas)
# 2 metros = 78.74 pulgadas, redondeamos a 79 para este ejemplo
jugadores_altos = df[df['height_inches'] > 79]
jugadores_altos.head()

# Múltiples condiciones (AND)
filtro = df[(df['height_inches'] > 79) & (df['weight'] > 220)]
filtro.head()

# Múltiples condiciones (OR)
filtro = df[(df['position'] == 'G') | (df['position'] == 'F')]
filtro.head()
```

Ordenar datos

```python
# Ordenar por una columna (ascendente por defecto)
df.sort_values('year_start').head()

# Ordenar descendente
df.sort_values('weight', ascending=False).head()

# Ordenar por múltiples columnas
df.sort_values(['position', 'height'], ascending=[True, False]).head()
```

**Transformación de columnas**

Crear columnas derivadas

```python
# Calcular años de carrera
df['duracion'] = df['year_end'] - df['year_start']
df[['name', 'year_start', 'year_end', 'duracion']].head()
```

Aplicar funciones con apply()

```python
# Clasificar carrera larga (> 10 años)
df['carrera_larga'] = df['duracion'].apply(lambda x: 1 if x > 10 else 0)
df['carrera_larga'].value_counts()
```

Eliminar columnas

```python
df = df.drop('carrera_larga', axis=1)
```

Renombrar columnas

```python
df = df.rename(columns={'name': 'nombre', 'position': 'posicion'})
```

**Agrupación y agregación**

La combinación de groupby() con funciones de agregación es una de las herramientas más potentes de Panda

```python
# Agrupar por posición y calcular promedios
promedios_posicion = df.groupby('posicion')[['height_inches', 'weight']].mean()
print(promedios_posicion)
```

Ordenar resultados agrupados

```python
# Top 5 posiciones con mayor peso promedio
promedios_posicion.sort_values('weight', ascending=False).head()
```

Múltiples agregaciones

```python
# Varias estadísticas a la vez
resumen = df.groupby('posicion')['height_inches'].agg(['mean', 'min', 'max', 'count'])
print(resumen)
```

**Visualizaciones con Pandas**

Pandas integra Matplotlib directamente. El método .plot() facilita la creación de gráficos

**Gráfico de barras (frecuencias)**

```python
# Frecuencia por posición (top 10)
df['posicion'].value_counts().head(10).plot(kind='bar', figsize=(10, 6))
plt.title('Distribución de Jugadores por Posición')
plt.xlabel('Posición')
plt.ylabel('Cantidad')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Histograma (distribución de alturas)

```python
df['height_inches'].plot(kind='hist', bins=20, figsize=(10, 6), edgecolor='black')
plt.title('Distribución de Alturas de Jugadores NBA')
plt.xlabel('Altura (pulgadas)')
plt.ylabel('Frecuencia')
plt.show()
```

**Gráfico de dispersión (relación entre variables)**

```python
df.plot(kind='scatter', x='height_inches', y='weight', figsize=(10, 6), alpha=0.5)
plt.title('Relación entre Altura y Peso')
plt.xlabel('Altura (pulgadas)')
plt.ylabel('Peso (libras)')
plt.show()
```

Boxplot (distribución por categoría)

```python
df.boxplot(column='weight', by='posicion', figsize=(12, 6))
plt.title('Distribución de Peso por Posición')
plt.suptitle('')  # Elimina título automático
plt.xlabel('Posición')
plt.ylabel('Peso (libras)')
plt.show()
```

Gráfico de líneas (series temporales)

```python
# Contar jugadores por año de inicio
por_año = df['year_start'].value_counts().sort_index()
por_año.plot(kind='line', figsize=(12, 6))
plt.title('Evolución de Jugadores que Ingresaron a la NBA')
plt.xlabel('Año')
plt.ylabel('Cantidad de Jugadores')
plt.show()
```

Gráfico de barras horizontales (top rankings)

```python
# Top 5 posiciones con mayor peso promedio
peso_promedio = df.groupby('posicion')['weight'].mean().sort_values(ascending=True).tail(5)
peso_promedio.plot(kind='barh', figsize=(10, 6), color='steelblue')
plt.title('Top 5 Posiciones por Peso Promedio')
plt.xlabel('Peso Promedio (libras)')
plt.ylabel('Posición')
plt.show()
```
